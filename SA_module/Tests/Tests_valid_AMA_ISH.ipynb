# Test the implementation of AMA indices on Ishigami function 

```python
import sys
sys.path.append('../')

from AMA_indices import ishigami, compute_stat_moments, Ama_indices
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Deterministic parameters values set in Dell'Oca et al., 2017

a = 5
b = 0.1

n_simu = int(5e5)
```

```python
# Random variables : x1, x2, x3 sampling from uniform distributions ([0,1])

x1 = np.random.uniform(size = n_simu)
x2 = np.random.uniform(size = n_simu)
x3 = np.random.uniform(size = n_simu)
```

```python
# Evaluate ISH function 
y = ishigami(x1, x2, x3, a, b)
# Compute unconditional statistical moments 
mm1, mm2 = compute_stat_moments(y)
```

Comparison with analytical values 

```python
print('E_emp[ISH] = '+str(mm1)[:5]+' (empirical (500 000 sim))')
print('E_an[ISH] = '+str(a/2)+' (analytical value a/2)')
```

```python
var_an = 1/2 + (a*a)/8 + b * np.power(np.pi,4) * (1/5 + (b*np.power(np.pi,4))/18)

print('V_emp[ISH] = '+str(mm2)[:6]+' (empirical (500 000 sim))')
print('V_an[ISH] = '+str(var_an)+' (analytical value)')
```

AMA indices 

```python
nb_bins = 50


AMAE_x1, AMAV_x1 = Ama_indices(nb_bins, x1, y, mm1, mm2)
AMAE_x2, AMAV_x2 = Ama_indices(nb_bins, x2, y, mm1, mm2)
AMAE_x3, AMAV_x3 = Ama_indices(nb_bins, x3, y, mm1, mm2)
```
