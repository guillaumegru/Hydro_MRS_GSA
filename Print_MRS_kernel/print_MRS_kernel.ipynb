# Print MRS Kernels computed with Samovar

```python
import sys
sys.path.append("../Phydrus_MRS_modules-py")

from MRS_load_interp_mod import loadMRSkernel, plotMRSkernel
import matplotlib.pyplot as plt
```

```python
info, model, depth_ref, pulse_ref, ker = loadMRSkernel('RMPTL_8sq40decl70_50m_high')
```

```python
q_numexp = [60.0, 144.6, 361.6, 903.9, 2260.0]
```

```python
plotMRSkernel(pulse_ref, depth_ref, ker, min(pulse_ref), max(pulse_ref), r'MRS sensitivity function $\kappa_{ref}$ $[nV.m^{-1}]$', fontsize = 15,savefig= True,  titlepng='MRS_kernel_fig', ylabel = 'z [m]', xlabel = 'q [A.ms]', cmap = 'hot')
plt.vlines(q_numexp, ymin=0, ymax=50, colors='grey',linestyles='--')
plt.scatter(q_numexp, max_sensitivities_depths, marker='x', color = 'grey')
plt.savefig('MRS_kernel_fig.png', dpi=300)
```
