## Compute unconditional mean and variance time series for all q

```python
nt = Y_E1.shape[1]
nparams = X.shape[1]
```

```python
mm1_E1 = np.zeros(nt)
mm1_E2 = np.zeros(nt) 
mm1_E3 = np.zeros(nt)
mm1_E4 = np.zeros(nt)
mm1_E5 = np.zeros(nt)
```

```python
mm2_E1 = np.zeros(nt)
mm2_E2 = np.zeros(nt) 
mm2_E3 = np.zeros(nt)
mm2_E4 = np.zeros(nt)
mm2_E5 = np.zeros(nt)
```

```python
# q1
for i in range(nt):
    mm1_temp, mm2_temp = compute_stat_moments(Y_E1[:,i])
    mm1_E1[i] = mm1_temp
    mm2_E1[i] = mm2_temp
```

```python
# q2
for i in range(nt):
    mm1_temp, mm2_temp = compute_stat_moments(Y_E2[:,i])
    mm1_E2[i] = mm1_temp
    mm2_E2[i] = mm2_temp
```

```python
# q4
for i in range(nt):
    mm1_temp, mm2_temp = compute_stat_moments(Y_E4[:,i])
    mm1_E4[i] = mm1_temp
    mm2_E4[i] = mm2_temp
```

```python
# q5
for i in range(nt):
    mm1_temp, mm2_temp = compute_stat_moments(Y_E5[:,i])
    mm1_E5[i] = mm1_temp
    mm2_E5[i] = mm2_temp
```
