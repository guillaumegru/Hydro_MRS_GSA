# Postprocess GSA of hydrogeophysical model (v5.6, With MRS undetectable water V2 and corrected geometry)

Histogram and pdf of signal MRS distributions 

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde
```

```python
#STEP2 : Retrieve simulation data
simfiles_dir = '../Hydro_MRS_GSA_numexp/Res_sim/'
fn_st = 'GSA5_5_'
n = '1'

# first set of simulation results
Y_E1_1 = np.loadtxt(simfiles_dir+fn_st+n+'_E1.csv', delimiter = ',')
Y_E2_1 = np.loadtxt(simfiles_dir+fn_st+n+'_E2.csv', delimiter = ',')
Y_E3_1 = np.loadtxt(simfiles_dir+fn_st+n+'_E3.csv', delimiter = ',')
Y_E4_1 = np.loadtxt(simfiles_dir+fn_st+n+'_E4.csv', delimiter = ',')
Y_E5_1 = np.loadtxt(simfiles_dir+fn_st+n+'_E5.csv', delimiter = ',')

X1 = np.loadtxt(simfiles_dir+fn_st+n+'_X.csv', delimiter = ',')


# second set of simulation results
n = '2'

Y_E1_2 = np.loadtxt(simfiles_dir+fn_st+n+'_E1.csv', delimiter = ',')
Y_E2_2 = np.loadtxt(simfiles_dir+fn_st+n+'_E2.csv', delimiter = ',')
Y_E3_2 = np.loadtxt(simfiles_dir+fn_st+n+'_E3.csv', delimiter = ',')
Y_E4_2 = np.loadtxt(simfiles_dir+fn_st+n+'_E4.csv', delimiter = ',')
Y_E5_2 = np.loadtxt(simfiles_dir+fn_st+n+'_E5.csv', delimiter = ',')

X2 = np.loadtxt(simfiles_dir+fn_st+n+'_X.csv', delimiter = ',')

n = '3'

Y_E1_3 = np.loadtxt(simfiles_dir+fn_st+n+'_E1.csv', delimiter = ',')
Y_E2_3 = np.loadtxt(simfiles_dir+fn_st+n+'_E2.csv', delimiter = ',')
Y_E3_3 = np.loadtxt(simfiles_dir+fn_st+n+'_E3.csv', delimiter = ',')
Y_E4_3 = np.loadtxt(simfiles_dir+fn_st+n+'_E4.csv', delimiter = ',')
Y_E5_3 = np.loadtxt(simfiles_dir+fn_st+n+'_E5.csv', delimiter = ',')

X3 = np.loadtxt(simfiles_dir+fn_st+n+'_X.csv', delimiter = ',')

n = '4'

Y_E1_4 = np.loadtxt(simfiles_dir+fn_st+n+'_E1.csv', delimiter = ',')
Y_E2_4 = np.loadtxt(simfiles_dir+fn_st+n+'_E2.csv', delimiter = ',')
Y_E3_4 = np.loadtxt(simfiles_dir+fn_st+n+'_E3.csv', delimiter = ',')
Y_E4_4 = np.loadtxt(simfiles_dir+fn_st+n+'_E4.csv', delimiter = ',')
Y_E5_4 = np.loadtxt(simfiles_dir+fn_st+n+'_E5.csv', delimiter = ',')

X4 = np.loadtxt(simfiles_dir+fn_st+n+'_X.csv', delimiter = ',')


X = np.concatenate((X1,X2,X3,X4))
Y_E1 = np.concatenate((Y_E1_1,Y_E1_2,Y_E1_3,Y_E1_4))
Y_E2 = np.concatenate((Y_E2_1,Y_E2_2,Y_E2_3,Y_E2_4))
Y_E3 = np.concatenate((Y_E3_1,Y_E3_2,Y_E3_3,Y_E3_4))
Y_E4 = np.concatenate((Y_E4_1,Y_E4_2,Y_E4_3,Y_E4_4))
Y_E5 = np.concatenate((Y_E5_1,Y_E5_2,Y_E5_3,Y_E5_4))
```

```python
Nsim_or = Y_E1.shape[0]
```

```python
nanidx = np.argwhere(np.isnan(Y_E1[:,0]))

Y_E1 = np.delete(Y_E1, nanidx, 0)
Y_E2 = np.delete(Y_E2, nanidx, 0)
Y_E3 = np.delete(Y_E3, nanidx, 0)
Y_E4 = np.delete(Y_E4, nanidx, 0)
Y_E5 = np.delete(Y_E5, nanidx, 0)
X = np.delete(X, nanidx, 0)
```

# Estimate signal pdf 

```python
linestyles = ['dashed', 'solid', 'dotted', 'dashdot', (0,(5,1))]

batlow10 = [(1,25,89), (16,  63,  96), (28,  90,  98), (60, 109,  86), (104, 123,  62), (157, 137,  43), (210, 147,  67), (248, 161, 123), (253, 183, 188), (250, 204, 250)]
batlow10_2 = []

for t in batlow10:
    r = t[0]/255
    g = t[1]/255
    b = t[2]/255
    batlow10_2.append((r,g,b))
```

```python
y_max_pdf = 0.023
fs = 15
```

## q1

```python
pdf_E1_kde = []
min_val = 1e3
max_val = 0

for t in t_toprint:
    min_val = min(min_val, np.min(Y_E1[:,t]))
    max_val = max(max_val, np.max(Y_E1[:,t]))
    pdf_E1_kde.append(gaussian_kde(Y_E1[:,t]))
```

```python
nsamples = 1e4
xx = np.arange(0, x_max, x_max/nsamples)
```

```python
pdf_E1_eval = []

for i in range(len(t_toprint)):
    pdf_E1_eval.append(pdf_E1_kde[i].evaluate(xx))
```

```python
plt.figure()
plt.title('q1 (60 A.ms)', fontsize= fs)
for k, t in enumerate(t_toprint):
    plt.plot(xx,pdf_E1_eval[k], label = 't = '+str(t)+' d', color = batlow10_2[2*k], linestyle = linestyles[k])
plt.xlabel(r'$E_0(q1) [nV]$', fontsize= fs)
plt.ylabel('Probability density', fontsize = fs)
plt.xlim((0,x_max))
plt.ylim((0,y_max_pdf))
plt.xticks(fontsize= fs)
plt.yticks(fontsize= fs)
plt.grid()
plt.legend(fontsize= fs)
plt.savefig('Figures/sig_hist_pdf/pdf_plot_E1.jpg', dpi=300, bbox_inches='tight')
```
